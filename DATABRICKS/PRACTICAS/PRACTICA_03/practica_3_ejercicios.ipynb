```python
from pyspark.sql import functions as F

ruta = "/Volumes/workspace/default/datos/productos.csv"
productos = spark.read.csv(ruta, header=True, inferSchema=True)

display(productos.limit(5))
```

```python
ruta = "/Volumes/workspace/default/datos/regiones.csv"
regiones = spark.read.csv(ruta, header=True, inferSchema=True)

display(regiones.limit(5))
```

```python
ruta = "/Volumes/workspace/default/datos/pedidos.csv"
pedidos = spark.read.csv(ruta, header=True, inferSchema=True)

display(pedidos.limit(5))
```

```python
'''
    EJERCICIO 1

   ¿Cuántos productos del catálogo están activos? 
'''

numero_productos_activos = productos.filter(F.col("activo") == "Sí").count()
print(f"En total hay {numero_productos_activos} productos activos")
```

```python
'''
    EJERCICIO 2

    Muestra las regiones con un objetivo de ventas superior a 5000 €, de mayor a menor objetivo.
'''

regiones_objetivo = regiones.select("region", "objetivo_ventas_2025").filter((F.col("objetivo_ventas_2025") > 5000)).orderBy(F.desc("objetivo_ventas_2025"))

display(regiones_objetivo)
```

```python
'''
    EJERCICIO 3
    
    ¿Cuántos pedidos hay de cada categoría? Ordénalos de más a menos
'''

display(pedidos.groupBy("categoria").count().orderBy(F.desc("count")))
```

```python
'''
    EJERCICIO 4

    Muestra los tres productos más caros, sin repetir. 
'''

display(productos.distinct().orderBy(F.desc("coste_unitario")).limit(3))
```
